## 2. Helper Functions

These are given so both approaches can share identical:
- `get_metrics`: macro precision/recall/F1 + accuracy.
- `get_stratified_splits`: builds **one** train/validation/test split that Approach 1 and Approach 2 will **both** reuse (this is what guarantees "both approaches are evaluated on the same held-out test set").
- `plot_loss_curve`: plots train vs. validation loss from a `Trainer`'s `log_history`.
- `plot_confusion_matrix_heatmap`: confusion matrix heatmap for either integer or string labels.
- `measure_latency_and_memory`: times batch-size-1 inference and logs peak CUDA memory, after a short warm-up (so the first, slower CUDA-kernel-compilation call doesn't skew results).

```python
def get_metrics(labels, preds):
    """
    Calculates macro-averaged evaluation metrics for classification.

    Args:
        labels: True ground truth labels.
        preds: Predicted labels from the model.

    Returns:
        A dictionary containing accuracy, precision, recall, and f1_score.
    """
    precision, recall, f1_score, _ = precision_recall_fscore_support(labels, preds, average='macro', zero_division=0)
    accuracy = accuracy_score(labels, preds)

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1_score": f1_score
    }


def get_stratified_splits(dataset_id, target_categories, seed=RANDOM_SEED,
                           train_frac=0.8, val_frac=0.1, test_frac=0.1):
    """
    Loads the Bitext dataset, filters to `target_categories`, and returns a single
    stratified train/validation/test DatasetDict
    """
    assert abs(train_frac + val_frac + test_frac - 1.0) < 1e-6

    label2id = {c: i for i, c in enumerate(target_categories)}
    id2label = {i: c for c, i in label2id.items()}

    dataset = load_dataset(dataset_id, split="train")
    dataset = dataset.filter(lambda x: x["category"] in target_categories)
    dataset = dataset.map(lambda x: {"label": label2id[x["category"]]})
    dataset = dataset.cast_column("label", ClassLabel(names=target_categories))

    holdout_frac = val_frac + test_frac
    train_holdout = dataset.train_test_split(test_size=holdout_frac, stratify_by_column="label", seed=seed)
    val_test = train_holdout["test"].train_test_split(
        test_size=test_frac / holdout_frac, stratify_by_column="label", seed=seed
    )

    splits = DatasetDict({
        "train": train_holdout["train"],
        "validation": val_test["train"],
        "test": val_test["test"],
    })
    return splits, label2id, id2label


def plot_loss_curve(log_history, title="Training Loss"):
    """Given a HF Trainer's `trainer.state.log_history`, plot train vs. eval loss per step."""
    train_steps, train_losses = [], []
    eval_steps, eval_losses = [], []
    for entry in log_history:
        if "loss" in entry and "eval_loss" not in entry:
            train_steps.append(entry.get("step", len(train_steps)))
            train_losses.append(entry["loss"])
        if "eval_loss" in entry:
            eval_steps.append(entry.get("step", len(eval_steps)))
            eval_losses.append(entry["eval_loss"])

    plt.figure(figsize=(8, 5))
    if train_losses:
        plt.plot(train_steps, train_losses, label="Train Loss", marker="o")
    if eval_losses:
        plt.plot(eval_steps, eval_losses, label="Validation Loss", marker="s")
    plt.xlabel("Step")
    plt.ylabel("Loss")
    plt.title(title)
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()


def plot_confusion_matrix_heatmap(labels, preds, class_names, title="Confusion Matrix"):
    """Confusion matrix heatmap. Works with integer labels or category-name strings."""
    use_names = isinstance(labels[0], str)
    cm = confusion_matrix(labels, preds, labels=class_names if use_names else list(range(len(class_names))))
    plt.figure(figsize=(9, 7))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=class_names, yticklabels=class_names)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(title)
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()


def measure_latency_and_memory(predict_one_fn, examples, warmup=3):
    """
    Calls predict_one_fn(example) once per example (batch_size=1, matching a
    realistic single-request serving scenario). Resets CUDA peak-memory stats
    after warm-up so the very first (slow, kernel-compiling) call doesn't
    skew latency, and the reported peak memory reflects only steady-state
    inference.
    """
    examples = list(examples)
    for example in examples[:warmup]:
        _ = predict_one_fn(example)

    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats()

    predictions = []
    start = time.time()
    for example in examples:
        predictions.append(predict_one_fn(example))
    end = time.time()

    avg_latency_ms = (end - start) / len(examples) * 1000
    peak_mem_mb = torch.cuda.max_memory_allocated() / (1024 ** 2) if torch.cuda.is_available() else 0.0
    return avg_latency_ms, peak_mem_mb, predictions
```

### 2. Fine-Tuning Encoder-Only Transformers

We fine-tune **all four candidate encoders** with the same recipe (AdamW + linear warmup, the `Trainer` default), pick the winner by **validation macro-F1** (the test set is never used for model selection), and then evaluate only that winner on the test set.


```python
# Candidate encoders -- all are trained with the same recipe; the best on VALIDATION macro-F1 is kept.
ENCODER_CANDIDATES = [
    "distilbert/distilbert-base-uncased",
    "google-bert/bert-base-uncased",
    "FacebookAI/roberta-base",
    "answerdotai/ModernBERT-base",
]

# Peek at how long a typical support message is (in tokens) before picking max_length
# padding every example out to the model's full context window wastes compute for no reason.
_peek_tok = AutoTokenizer.from_pretrained(ENCODER_CANDIDATES[0])
_sample_lengths = [len(_peek_tok.encode(t)) for t in splits["train"]["instruction"][:2000]]
print(f"Instruction token length -- mean: {np.mean(_sample_lengths):.1f}, "
      f"95th pct: {np.percentile(_sample_lengths, 95):.0f}, max: {max(_sample_lengths)}")

# Messages are short (max is well under 64 tokens), so 64 covers ~100% without truncation.
ENC_MAX_LEN = 64
```

```python
def compute_metrics(eval_pred):
    """Unpack eval_pred into (logits, labels), argmax the logits (axis=-1), reuse get_metrics."""
    ## YOUR CODE STARTS HERE
    logits, labels = eval_pred
    if isinstance(logits, tuple):
        logits = logits[0]
    preds = np.argmax(logits, axis=-1)
    ## YOUR CODE ENDS HERE
    return get_metrics(labels, preds)


def train_encoder(model_id):
    tok = AutoTokenizer.from_pretrained(model_id)
    def tokenize_function(examples):
        return tok(examples["instruction"], truncation=True, max_length=ENC_MAX_LEN)
    tokenized = splits.map(tokenize_function, batched=True)
    model = AutoModelForSequenceClassification.from_pretrained(
        model_id, num_labels=NUM_LABELS, id2label=id2label, label2id=label2id,
    ).to(device)

    args = TrainingArguments(
        output_dir=f"./enc_results/{model_id.split('/')[-1]}",
        learning_rate=3e-5,                  # AdamW (Trainer default optimizer)
        lr_scheduler_type="linear",          # linear decay ...
        warmup_ratio=0.1,                    # ... with linear warmup over the first 10% of steps
        weight_decay=0.01,
        per_device_train_batch_size=32,
        per_device_eval_batch_size=64,
        num_train_epochs=3,
        eval_strategy="epoch",
        save_strategy="epoch",
        save_total_limit=1,
        load_best_model_at_end=True,
        metric_for_best_model="f1_score",
        logging_strategy="steps",
        logging_steps=50,
        fp16=torch.cuda.is_available(),
        seed=RANDOM_SEED,
        report_to="none",
    )
    trainer = Trainer(
        model=model,
        args=args,
        train_dataset=tokenized["train"],
        eval_dataset=tokenized["validation"],
        processing_class=tok,
        data_collator=DataCollatorWithPadding(tokenizer=tok),
        compute_metrics=compute_metrics,
    )
    t0 = time.time()
    trainer.train()
    val = trainer.evaluate()
    val["train_minutes"] = (time.time() - t0) / 60
    return trainer, tok, val


sweep_results = {}
best_val_f1, enc_trainer, enc_tokenizer, encoder_model_id = -1, None, None, None
for mid in ENCODER_CANDIDATES:
    print(f"\n==================== Training {mid} ====================")
    try:
        trainer, tok, val = train_encoder(mid)
    except Exception as e:                      # e.g. an old transformers without ModernBERT support
        print(f"Skipping {mid}: {type(e).__name__}: {e}")
        continue
    sweep_results[mid] = val
    print(f"{mid}: val macro-F1 = {val['eval_f1_score']:.4f}")
    if val["eval_f1_score"] > best_val_f1:
        if enc_trainer is not None:            # free the previous best
            del enc_trainer.model, enc_trainer
        best_val_f1, enc_trainer, enc_tokenizer, encoder_model_id = val["eval_f1_score"], trainer, tok, mid
    else:
        del trainer.model, trainer
    gc.collect(); torch.cuda.empty_cache()

enc_model = enc_trainer.model
sweep_df = pd.DataFrame(sweep_results).T[["eval_accuracy", "eval_precision", "eval_recall", "eval_f1_score", "train_minutes"]]
print("\nValidation results of the encoder sweep (model selection uses VALIDATION only):")
display(sweep_df.sort_values("eval_f1_score", ascending=False))
print(f"\nSelected encoder: {encoder_model_id} (val macro-F1 {best_val_f1:.4f})")

# Save the selected encoder -- W15 exports this to ONNX and serves it.
enc_trainer.save_model(f"{ARTIFACT_DIR}/router_encoder")
enc_tokenizer.save_pretrained(f"{ARTIFACT_DIR}/router_encoder")
```

```python
def predict_one_encoder(example):
    """Tokenize a single instruction, run a forward pass, return the argmax label id."""
    inputs = enc_tokenizer(
        example["instruction"], return_tensors="pt", truncation=True, max_length=ENC_MAX_LEN
    ).to(device)
    with torch.no_grad():
        logits = enc_model(**inputs).logits
    return int(torch.argmax(logits, dim=-1).item())

enc_model.eval()
enc_latency, enc_peak_mem, enc_predictions = measure_latency_and_memory(
    predict_one_encoder, splits["test"]
)
enc_labels = splits["test"]["label"]

enc_metric_dict = get_metrics(enc_labels, enc_predictions)
enc_acc, enc_prec, enc_rec, enc_f1 = (enc_metric_dict["accuracy"], enc_metric_dict["precision"],
                                       enc_metric_dict["recall"], enc_metric_dict["f1_score"])

print(f"Encoder Macro-F1: {enc_f1:.4f}")
print(f"Encoder Latency: {enc_latency:.2f} ms/query")
print(f"Encoder Peak Memory: {enc_peak_mem:.2f} MB")
```

### 1. Model & Tokenizer Setup

```python
# Qwen2.5-0.5B-Instruct + LoRA fits comfortably in 8 GB in bf16, so we do plain LoRA (no 4-bit).
# Set DECODER_MODEL_ID to "Qwen/Qwen2.5-1.5B-Instruct" and USE_4BIT=True to try the larger model (QLoRA).
decoder_model_id = "Qwen/Qwen2.5-0.5B-Instruct"
USE_LORA = True
USE_4BIT = False   # QLoRA via bitsandbytes -- only needed for the 1.5B model on 8 GB

dec_tokenizer = AutoTokenizer.from_pretrained(decoder_model_id, trust_remote_code=True)
# Some causal LMs ship without a pad token
# set one if missing.
if dec_tokenizer.pad_token is None:
    dec_tokenizer.pad_token = dec_tokenizer.eos_token
dec_tokenizer.padding_side = "left"   # correct side for batched generation with decoder-only models

compute_dtype = torch.bfloat16 if (torch.cuda.is_available() and torch.cuda.is_bf16_supported()) else torch.float16

if USE_LORA and USE_4BIT:
    # QLoRA: load the base model in 4-bit (bitsandbytes) and only train LoRA adapters on top.
    bnb_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_compute_dtype=compute_dtype,
        bnb_4bit_use_double_quant=True,
    )
    dec_model = AutoModelForCausalLM.from_pretrained(
        decoder_model_id, quantization_config=bnb_config, device_map="auto", trust_remote_code=True,
    )
    dec_model = prepare_model_for_kbit_training(dec_model)
else:
    # bf16 weights; with USE_LORA only the adapters are trained (frozen base).
    dec_model = AutoModelForCausalLM.from_pretrained(
        decoder_model_id, dtype=compute_dtype, device_map="auto", trust_remote_code=True,
    )

dec_model.config.use_cache = False   # required before training with gradient checkpointing
```

### 2. Loading & Formatting the Dataset

Reuse the **same** `splits` `DatasetDict` built in Approach 1's Data Loading step -- do **not**
call `load_dataset` again here. This is what keeps the two approaches' test sets identical.

```python
print(splits)
print(splits["train"][0])
```

```python
SYS_PROMPT = (
    "You are an AI assistant trained for text classification; given any user "
    f"query, choose the single closest category from this fixed set: {TARGET_CATEGORIES}; "
    "output only the category name as the final answer without explanation or extra words."
)

def build_prompt(instruction_text):
    """Render the system + user turns with the model's own chat template."""
    messages = [
        {"role": "system", "content": SYS_PROMPT},
        {"role": "user", "content": instruction_text},
    ]
    return dec_tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

def format_example(example):
    """
    1. Build the chat-templated prompt (system + user turn).
    2. The target is example["category"]
    3. Append the EOS token after the label so the model learns *where to stop*.
    4. Return {"input_text": prompt + target + eos}.
    We also keep "prompt"/"completion" so SFTTrainer computes the loss on the label only.
    """
    ## YOUR CODE STARTS HERE
    prompt = build_prompt(example["instruction"])
    target = example["category"]
    ## YOUR CODE ENDS HERE

    return {"input_text": prompt + target + dec_tokenizer.eos_token,
            "prompt": prompt, "completion": target + dec_tokenizer.eos_token}

# Apply it across every split, keeping the original instruction text around
formatted_splits = splits.map(
    lambda ex: {**format_example(ex), "original_text": ex["instruction"]}
)
print(formatted_splits["train"][0]["input_text"])
```

### 3. Baseline Check: Zero-Shot Inference (Before Fine-Tuning)

```python
# Given: greedy-decodes up to max_new_tokens tokens (a label is one word, so 5 is generous headroom).
def get_outputs(model, inputs, max_new_tokens=5):
    return model.generate(
        input_ids=inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_new_tokens=max_new_tokens,
        repetition_penalty=1.1,
        eos_token_id=dec_tokenizer.eos_token_id,
        pad_token_id=dec_tokenizer.pad_token_id,
        do_sample=False,
    )

def normalize_decoder_output(text, valid_categories=None):
    """Lowercase + strip everything but letters, then snap to a known category if the
    raw text contains one as a substring (handles the model echoing extra words/punctuation
    around the label). Falls back to the cleaned string, which will simply score as an
    incorrect prediction rather than crash the metric computation."""
    valid_categories = valid_categories or [c.lower() for c in TARGET_CATEGORIES]
    cleaned = re.sub(r"[^a-z]", "", text.lower())
    for cat in valid_categories:
        if cat in cleaned:
            return cat
    return cleaned

# TODO: turn a raw generation into a clean predicted category string
def get_decoder_predictions(dataset, model):
    """For every row: rebuild the chat prompt from row["original_text"], generate,
    decode only the newly-generated tokens, and normalise. Returns (labels, predictions),
    both lists of lowercased strings."""
    labels, predictions = [], []
    for row in dataset:
        prompt = build_prompt(row["original_text"])
        inputs = dec_tokenizer(prompt, return_tensors="pt").to(model.device)
        output_ids = get_outputs(model, inputs)
        new_tokens = output_ids[0][inputs["input_ids"].shape[1]:]
        decoded = dec_tokenizer.decode(new_tokens, skip_special_tokens=True)
        predictions.append(normalize_decoder_output(decoded))
        labels.append(row["category"].lower())
    return labels, predictions

# Zero-shot baseline on a fixed 500-example sample of the test set (saves ~2 min; the full test set is used after fine-tuning)
_zs_sample = formatted_splits["test"].shuffle(seed=RANDOM_SEED).select(range(500))
dec_labels, dec_preds = get_decoder_predictions(_zs_sample, dec_model)
print("Zero-shot (pre-fine-tuning) baseline:", get_metrics(dec_labels, dec_preds))
```

```python
_dec_lengths = [len(dec_tokenizer.encode(t)) for t in formatted_splits["train"]["input_text"][:1000]]
print(f"Formatted prompt+label token length -- mean: {np.mean(_dec_lengths):.1f}, "
      f"95th pct: {np.percentile(_dec_lengths, 95):.0f}, max: {max(_dec_lengths)}")

# System prompt (~75 tokens) + longest instruction + label + EOS; round up with headroom.
DEC_MAX_LEN = int(max(_dec_lengths) + 16)
print("DEC_MAX_LEN =", DEC_MAX_LEN)

def tokenize_function(examples):
    """Tokenize examples["input_text"] with truncation to a fixed max_length."""
    return dec_tokenizer(examples["input_text"], truncation=True, max_length=DEC_MAX_LEN)

dec_tokenized = formatted_splits.map(tokenize_function, batched=True)

# SFTTrainer is given prompt/completion pairs so the loss is computed on the label tokens only
# (the system prompt is identical for every row -- training on it would just add noise to the loss curve).
dec_train_ds = formatted_splits["train"].select_columns(["prompt", "completion"])
dec_val_ds = formatted_splits["validation"].shuffle(seed=RANDOM_SEED).select(range(600)).select_columns(["prompt", "completion"])
```

### 5. LoRA Config & Training Arguments

```python
peft_config = None
if USE_LORA:
    peft_config = LoraConfig(
        r=16,                 # rank 16 is plenty for an 11-way routing head
        lora_alpha=32,
        lora_dropout=0.05,
        bias="none",
        task_type="CAUSAL_LM",
        target_modules="all-linear",
    )

dec_train_args = SFTConfig(
    output_dir=f"{decoder_model_id.split('/')[-1]}-router-finetuned/",
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    gradient_accumulation_steps=4,          # effective batch 64
    optim="paged_adamw_32bit" if USE_4BIT else "adamw_torch",
    learning_rate=2e-4 if USE_LORA else 2e-5,
    max_grad_norm=0.3,
    num_train_epochs=1,                     # ~336 optimizer steps; the label is easy to learn
    warmup_ratio=0.1,
    lr_scheduler_type="cosine",
    logging_strategy="steps",
    logging_steps=10,
    eval_strategy="steps",     # NOTE: required so eval_loss appears in log_history for the loss-curve plot
    eval_steps=25,
    save_strategy="no",
    bf16=compute_dtype == torch.bfloat16,
    fp16=compute_dtype == torch.float16,
    max_length=DEC_MAX_LEN,
    completion_only_loss=True,
    seed=RANDOM_SEED,
    report_to="none",
)
```

### 6. Model Training

```python
## Supervised Fine-Tuning (SFT)
dec_trainer = SFTTrainer(
    model=dec_model,
    train_dataset=dec_train_ds,
    eval_dataset=dec_val_ds,
    peft_config=peft_config,
    args=dec_train_args,
    processing_class=dec_tokenizer,
)
if USE_LORA:
    dec_trainer.model.print_trainable_parameters()

print("Training Decoder Model...")
dec_trainer.train()
dec_trainer.model.config.use_cache = True
dec_trainer.save_model(f"{ARTIFACT_DIR}/router_decoder_lora")
```

```python
chit_chat = ["Hey! How's it going?", "what's the weather like today", "tell me a joke", "thanks, bye!"]
rows = []
for msg in chit_chat:
    inputs = enc_tokenizer(msg, return_tensors="pt", truncation=True, max_length=ENC_MAX_LEN).to(device)
    with torch.no_grad():
        probs = softmax(enc_model(**inputs).logits, dim=-1)[0]
    conf, idx = probs.max().item(), int(probs.argmax())
    dec_out = predict_one_decoder({"original_text": msg})
    rows.append({"message": msg, "encoder_label": id2label[idx], "encoder_confidence": round(conf, 3),
                 "decoder_label": dec_out})
# Reference: typical encoder confidence on real in-domain test messages
_in = [softmax(enc_model(**enc_tokenizer(splits["test"][i]["instruction"], return_tensors="pt").to(device)).logits, -1).max().item()
       for i in range(200)]
print(f"Median encoder confidence on in-domain test messages: {np.median(_in):.3f}")
pd.DataFrame(rows)
```

## 9. Reflection

### Q1: On what transfers from pretraining

**What pretraining contributed.** Neither model has seen a support ticket, but both arrived knowing English: subword vocabularies that already cover informal spellings ("pls", "acct"), contextual representations where *"charged twice"*, *"double billed"* and *"two payments"* land close together, and syntax that separates *what* is being asked from filler ("hi, I was wondering if you could…"). That is why the encoders get strong validation macro-F1 after one epoch on ~21k short messages, and why the decoder's **zero-shot** baseline is already well above chance (1/11 ≈ 9%): the instruction-tuned Qwen can map "refund" → REFUND from label names alone.

**What fine-tuning had to teach from scratch.** (1) *The label space and its boundaries.* The 11 agents are a business taxonomy, not a linguistic one. Pretraining does not know that "cancel my order" belongs to ORDER, whereas "what is the fee to cancel" belongs to CANCEL, or that a "refund for my payment" should go to REFUND and not PAYMENT. Most of the remaining confusion-matrix errors sit on exactly these boundaries (ORDER↔CANCEL, PAYMENT↔REFUND↔INVOICE, SHIPPING↔DELIVERY). (2) *The output interface.* For the encoder, the classification head on [CLS] is randomly initialised: the map from representation → 11 logits is learned entirely during fine-tuning. For the decoder, fine-tuning teaches the format: emit exactly one uppercase agent name and then EOS, instead of chatty text. That is why the invalid-generation count drops to ~0 after SFT. (3) *Dataset-specific phrasing.* Bitext templates such as `{{Order Number}}` placeholders are artefacts that the models learn to ignore.

### Q2: On Handling Chit-Chat Queries
If a user sends *"Hey! How’s it going?"*:

**Encoder (softmax over 11 classes).** The architecture has no "none of the above" option. "Hey! How's it going?" is forced into one of the 11 agents, typically CONTACT or FEEDBACK (the most "conversational" classes), and the chit-chat probe cell above shows how confident it is. Its failure is *silent*: the label looks valid. Its one useful signal is a **max-softmax probability** that is usually much lower than on in-domain messages (compare it with the median printed above), so the failure can be detected with a threshold.

**Decoder SLM (generation).** After SFT it has also been trained to always emit one of the 11 names, so it too will usually output something like CONTACT. Before fine-tuning it might have replied "I'm doing well, thanks!", which `normalize_decoder_output` turns into an invalid label. After fine-tuning that escape hatch is largely trained away, and generation gives no calibrated confidence without extra work (e.g. reading the first-token logprobs).

**Which fails most gracefully?** The encoder: its error is cheap (~ms), deterministic and comes with a usable confidence score that lets us abstain. The decoder is as likely to misroute and gives no native signal that it is unsure.

**What I would add in production.** (1) A **12th class `OTHER/CHITCHAT`**, trained on chit-chat, greetings, off-topic and multi-intent examples (the Bitext set has none, so we would source them from logs or synthesise them). (2) A **confidence gate**: if max-softmax < τ (τ tuned on validation to e.g. 95% in-domain recall), do not dispatch. Instead, send the message to a lightweight "greeter / clarifier" agent that replies politely and asks what they need help with. (3) An **OOD detector** on the embedding (e.g. Mahalanobis or kNN distance to training centroids) as a second signal. (4) Monitoring: log the share of low-confidence and OTHER routings per day, and alert when it rises. A rise means either more chit-chat or real drift in customer phrasing.
